```python
def mult_three(x):
    '''return 3 times the parameter x'''

    result = 3 * x
    return result # produce a reuslt from this function

# test case:
r = mult_three(14)
print("r =", r)

print('mult_three(0.25) returned', mult_three(0.25))
print('mult_three("abc") returned', mult_three("abc"))
```

Discuss/demonstrate: the difference between print and return

```python
def function_a(x):
    result = x + 1
    print(result) # prints an output

def function_b(x):
    result = x + 1
    return result # returns a result

# call the function:
# function_a(3) # (see the printout)
# function_b(3) # (see the printout)

# try to assign the result to a variable:
value_a = function_a(3)
print(value_a)
value_b = function_b(3)
print(value_b)

# c = value_a + 1 # use the variable in another calculation
d = value_b + 1 # use the result of function_b in another calculation
print("d =", d)
```

Functions with list as parameters

```python
def calc_average(values):
    '''return the mean of a list of integers'''

    return sum(values) / len(values)

# test code:
numbers = [8,6,7,5,3,0,9]
print('calc_average(numbers) =', calc_average(numbers))
```

Random selection from a list of strings

```python
fruits = ['apples', 'bananas', 'pears', 'strawberries', 'nectaries', 'grapes']
# random.choice will select one item from a list
selected = random.choice(fruits)
print("selected =", selected)

# how does random.choice work?
# find the length of the list
list_len = len(fruits)
# pick a random index in the range (0, length-1)
index = random.randint(0, list_len-1)
# list indexing to extract result
print("we picked", fruits[index])

```

Passing a function to another function.

```python
def execute(f, data):
    '''execute the function f (to be specified later) on the list data'''
    
    return f(data)

data = [8,6,7,5,3,0,9]
print(execute(sum, data))

```
