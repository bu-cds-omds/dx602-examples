Using dictionary as an accumulator

```python
import random
numbers = [random.randint(1,10) for i in range(25)]
print("numbers = ", numbers)

# goal: count occurrences of some values
# set up an accumulator variable:
counts = {} # an empty dictionary

# use a loop to process the list of numbers
for n in numbers:

    # have we seen n before?
    if n in counts: # check if n is a valid key in numbers:
        # accumulate the counts for each value
        counts[n] = counts[n] + 1
    else:
        # insert this key n for the first time:
        counts[n] = 1 # first time we've seen this key!

# print out the accumulated results
print("counts = ", counts)
#for k in counts.keys():
# for k in sorted(counts.keys()):
for k in sorted(counts.keys(), reverse=True):
    # print("k=", k, k * "*") # k is the key, not the count for the key!
    print("k=", k, counts[k] * "*") #counts[k] is the value (count) associated with k

```

```python
## same example with a string instead of random numbers
song = "the long and winding road to your heart"
print("song = ", song)

# goal: count occurrences of each letter!
# set up an accumulator variable:
counts = {} # an empty dictionary

# use a loop to process the characters in song one at a time
for c in song:

    # have we seen c before?
    if c in counts: # check if c is a valid key in counts:
        # accumulate the counts for each value
        counts[c] = counts[c] + 1
    else:
        # insert this key c for the first time:
        counts[c] = 1 # first time we've seen this key!

# print out the accumulated results
print("counts = ", counts)
#for k in counts.keys():
# lambda k: counts[k] tells sorted to use the values as the element to sort on
for k in sorted(counts.keys(), key=lambda k: counts[k], reverse=True):
# for k in sorted(counts.values(), reverse=True): # can't sort by values
    print(k, counts[k] * "*") #counts[k] is the value (count) associated with k

# idea: use a minimum threshhold limit
# show_keys = [k for k in counts if counts[k] >= 3]
# for k in sorted(show_keys, key=lambda k: counts[k], reverse=True):
    # print("k=", k, k * "*") # k is the key, not the count for the key!
    # print(k, counts[k] * "*") #counts[k] is the value (count) associated with k

# print("counts.values()," , counts.values())
# print("max(counts),", max(counts))
```

What about initializing the dictionary with all possible valid keys?

```python
## same example with a string instead of random numbers
song = "the long and winding road to your heart"
print("song = ", song)

character_set = set(song)
print("character_set = ", character_set)
# goal: count occurrences of each letter!
# set up an accumulator variable:
counts = { c: 0 for c in character_set} # dictionary comprehension syntax
print("counts =", counts)

# use a loop to process the characters in song one at a time
for c in song:

    # incremene the count for each character c
    counts[c] = counts[c] + 1


for k in sorted(counts.keys(), key=lambda k: counts[k], reverse=True):

    print(k, counts[k] * "*") #counts[k] is the value (count) associated with k
```

Sorting:
<li> dictionaries cannot be sorted
<li> discuss list of `dict`s
<li> lists of key/value pairs can be sorted using the `list`.sort() method


```python
# data is a **list** containing dictionaries
data = [{'age': 22, 'height_cm': 183, 'weight_kg': 51},
         {'age': 62, 'height_cm': 168, 'weight_kg': 57}, 
         {'age': 49, 'height_cm': 129, 'weight_kg': 98}]

# sort this list by the height_cm, reassign to data
data = sorted(data, key=lambda d: d['height_cm'])

for d in data:
    print(d)
```

Additional dictionary practice

```python
# combine two dictionaries
dict1 = {'a': 5, 'b':1}
dict2 = {'b': 4, 'c':6}
# print(dict1 + dict2) # TypeError: unsupported operand type(s) for +: 'dict' and 'dict'

# find all unique keys from both dictionaries:
all_keys = set(list(dict1.keys()) + list(dict2.keys()))
print("all_keys", all_keys)

# accumulator pattern:
dict_combined = {k:0 for k in all_keys} # start with counts of 0

# iterate over all keys:
for k in all_keys:
    # still need to check if the key exists in each source dict
    if k in dict1:
        dict_combined[k] = dict_combined[k] + dict1[k] 
    if k in dict2:
        dict_combined[k] = dict_combined[k] + dict2[k] 
print("dict_combined = ", dict_combined)
```
