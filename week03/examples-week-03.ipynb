Write a function named `traffic_fine(speed, limit)` that:

<li> Accepts two integers: speed (vehicle speed in km/h) and limit (speed limit in km/h).
<li> Returns the appropriate fine string based on how much the speed exceeds the limit:
<li> If speed is less than or equal to limit: return "No fine".
<li> If speed is 1–20 km/h over the limit: return "Fine: $100".
<li> If speed is 21–40 km/h over the limit: return "Fine: $250".
<li> If speed is more than 40 km/h over the limit: return "Fine: $500 and license suspension".
<li> If either input is negative, return "Invalid input".

```python
def traffic_fine(speed, limit):
    '''
    Accepts two integers: speed (vehicle speed in km/h) and limit (speed limit in km/h).
    Returns the appropriate fine string based on how much the speed exceeds the limit:
    If speed is less than or equal to limit: return "No fine".
    If speed is 1–20 km/h over the limit: return "Fine: $100".
    If speed is 21–40 km/h over the limit: return "Fine: $250".
    If speed is more than 40 km/h over the limit: return "Fine: $500 and license suspension".
    If either input is negative, return "Invalid input".
    '''
    # put most specific cases first:
    if speed < 0:
        return "invalid speed"
    if limit < 0:
        return "invalid limit"
    if (speed - limit) <= 20:
        return "Fine: $100"
    if (speed - limit) <= 40:
        return "Fine: $250"  
    if (speed - limit) > 40:
        return "Fine: $500; license suspension"   
    # most general case at the bottom
    if speed <= limit:
        # print("no fine")
        return "no fine"
# test cases:
print(f"traffic_fine(45, 50), {traffic_fine(45, 50)}")
print(f"traffic_fine(55, 50), {traffic_fine(55, 50)}")
print(f"traffic_fine(75, 50), {traffic_fine(75, 50)}")
print(f"traffic_fine(97.7, 50), {traffic_fine(97.7, 50)}")
print(f"traffic_fine(-15, 50), {traffic_fine(-15, 50)}")
```

**Optimization using a list of lists**
<li> start with a list of options
<li> need a function to "score" each option
<li> build list of sublists
<li> pick the best one using max()

```python
def num_vowels(word):
    '''Return the number of vowels ('aeiou') in a word.'''
    # vowel_count = [c in 'aeiou' for c in word] # list of booleans
    # vowel_count = [1 for c in word if c in 'aeiouAEIOU'] # list of 1s
    vowel_count = [c in 'aeiouAEIOU' for c in word ] # list of booleans
    return sum(vowel_count)

# test case:
# print(num_vowels('strawberry'))
fruits = ['apple', 'banana', 'cherry', 'strawberry', 'lime', 'cononut']
# which fruit has the most vowels?

# 1: score each word in the list (# vowels)
# score_words = [ (w, num_vowels(w)) for w in fruits] # incorrect!
score_words = [ (num_vowels(w), w) for w in fruits]
print('score_words', score_words)
# 2: max: pick the highest score
best_pair = max(score_words)
print('best_pair =', best_pair)
# 3: extract the word corresponding to that highest score
best_word = best_pair[1] # indexing
print("The most vowels is", best_word)
```

Another example: which number from 1 to 100 has the most factors?

```python
def factors(n):
    '''Return the list of factors of n.'''
    return [ i for i in range(1, n//2+1) if n % i == 0]

def num_factors(n):
    '''Return the number of factors of n.'''
    f = factors(n)
    return len(f)

def most_factors(numbers):
    '''return the number from the list numbers with the most factors'''

    # 1 - score the options:    
    scored_options = [(num_factors(x), x) for x in numbers]
    # print(scored_options)
    # 2- pick the max:
    best = max(scored_options)
    # print("best = ", best)
    # 3- use indexing to extract the result
    return best[1] # the number with most factos

# print("factors(20) =", factors(20))
# print("num_factors(20) =", num_factors(20))
print("most_factors(range(10))", most_factors(range(10)))
print("most_factors(range(100))", most_factors(range(100)))
print(factors(96))
```
