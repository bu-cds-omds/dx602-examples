Accumulator pattern: building a value iteratively with a loop

```python
def add_up(numbers):
    '''add up some numbers; return the sum'''

    # declare an "accumulator" variable:
    total = 0

    # a loop to process a list of numbers:
    for i in range(len(numbers)):

        # add the ith element into total
        print('next number:', numbers[i])
        # total = total + numbers[i] 
        total += numbers[i] # shortcut operator
        print('total:', total)

    # return the accumulated result:
    return total

numbers = [2,1,9]
print('add_up(numbers) =', add_up(numbers))

# def add_up(numbers):
#     '''short version!'''
#     return sum(numbers)

# print('add_up(numbers) =', add_up(numbers))
```

Accumulator example: integer result<br> 
Computing factorials

```python
# factorial: n! = n * (n-1) * (n-2) * ... * 1
def factorial(n):
    '''return the factorial of n'''

    # declare an "accumulator" variable:
    fact = 1

    # a loop to process a sequence:
    for i in range(1, n+1):
        #print("i =", i)
        # do our one step with one element of the sequence
        fact = fact * i

    # return the accumulated result:
    return fact

# testing!!
print('factorial(5) =', factorial(5))
```

Accumulating a list<br>
Example: reciprocal numbers



```python
def reciprocal_numbers(n):
    '''return a list of reciprocal numbers up to n.
    reciprocal: 1/i'''

    # declare an "accumulator" variable:
    results = [] # start with an empty list

    # a loop to process a sequence:
    for i in range(n):

        # do our one step with one element of the sequence
        value = 1 / (i + 1)
        # append the value into our results
        results.append(value)

    # return the accumulated result:
    return results

# testing!
print('reciprocal_numbers(5) =', reciprocal_numbers(5))
```

Accumulating a list<br>
Example: Fibonacci numbers



```python
def fib(n):
    '''generate the first n Fibonacci numbers:
    [0, 1, 1, 2, 3, 5, 8, ...]'''

    # declare accumulator variable
    results = [0, 1]

    # use a loop to iterate over a sequence
    for i in range(n-2):

        # print('results =', results)
        # compute next value as sum of last 2 elements
        v = results[-2] + results[-1]
        results.append(v) # add to results

    # return results
    return results

# test case
print("fib(20) =", fib(20))
```

ASCII Codes, Converstion

```python
# reminder: ord(ch) -> number, chr(x) -> character
# print("ord('q')=",ord('q'))
# print('chr(100)=',chr(100))

# take a string, return list of ASCII codes
def str_to_ascii(s):
    '''process a string, return list of ASCII codes'''

    # declare result variable:
    results = []

    # iterate over each character in a string s:
    for ch in s:

        # append a value to the results:
        results.append( ord(ch) )

    # return result:
    return results

print("str_to_ascii('i love chocolate') =", str_to_ascii('i love chocolate'))
```

Problem 12: Caesar Cipher

```python
# shift characters by 1 position:
s = 'chocolate' # plain text
# s = 'dipdpmbuf' # ciphered text

# a string accumulator
result = '' 

# iterate over each character in the string
for ch in s:
    # take the numeric value of ch
    n = ord(ch)
    # add 1
    n = n + 1
    # convert back to character
    ch_mod = chr(n)

    # append to the accumulator
    result += ch_mod

print('result =', result)
```
