Numpy arrays

```python
a = np.array([1,2,3,4,5])
b = np.array(range(10,5,-1))
print(f'a = {a}')
print(f'b = {b}')

# element-wise operation
# c = a * 10 + 4.2 + b
# c = a * b # element-wise multiplication of 2 arrays ( a[0] * b[0] , a[1] * a[1])
c = a @ b # dot-product multiplication (a[0] * b[0] + a[1] * b[1] + ... )
# c = np.dot(a, b)
print(f'c = {c}')

# aggregate operations
print(f'sum = {a.sum()}')
print(f'mean = {a.mean()}')
print(f'std = {a.std()}')
```

The shape of an array (i.e., dimensions)

```python
a = np.array([10, 20, 30, 40])
b = np.array([40, 50, 60, 70])
c = b.reshape(4, 1) # column vector
# print(f'a = {a}')
# print(f'b = {b}')
# print(f'c = \n{c}')

# b[0] = 99 # changing one element in b
# print('after b[0] = 99')
print(f'a = {a}')
print(f'b = {b}')
print(f'c = \n{c}')

# create product of a and b:
e = a * b.reshape(4, 1) # element-wise product; producing a 4x4 result
print(f'e = {e}')
print(f'e.sum() = {e.sum()}')
print(f'e.sum(axis=0) = {e.sum(axis=0)}') # column-wise sum
# print(f'e.sum(axis=1) = {e.sum(axis=1)}') # row-wise sum
print(f'e.sum(axis=1).reshape(4,1) = \n{e.sum(axis=1).reshape(4,1)}') # row-wise sum
```

Transpose of an array

```python
# use nested list comprehension to create a 5-row, 3-col array
values = np.array([[ i + j for i in range(0, 3)]
                  for j in range(0,5)])
print(f'values= \n{values}')
print(f'values.reshape(3, 5) = \n{values.reshape(3, 5)}')
# print(f'values.reshape(2, 5) = \n{values.reshape(2, 5)}')
print(f'values.transpose() =\n {values.transpose()}')




```
