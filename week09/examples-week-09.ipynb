```python
# general import statements
import matplotlib.pyplot as plt
import numpy as np

# set a constant value
BAR_CHART_COL_WIDTH = 0.8
```

Basic plots...

```python
numbers = [3,1,4,5,9]
# plt.plot(numbers)
x = [1,2,3,4,5]
plt.plot(x, numbers)
plt.ylabel('values')
plt.xlabel('numbers')
plt.title('we like pie!')  # method call  
# dont this:
# plt.title = "we like pie!" # re-assign as a string
```

Plotting multiple things together

```python
plt.plot(x, numbers, color="blue", marker="*")
plt.bar(x=x, height=numbers, color="#9400D3")

```

Some waves

```python
# generate some x and y values
x = np.arange(0, 10, 0.1)
# print(x)
y = np.sin(x)
plt.plot(x, y)
plt.title("catch a wave and you're sitting on top of the world!")

```

Two waves...

```python
# generate some x and y values
x = np.arange(0, 10, 0.1)
# print(x)
y1 = np.sin(x)
y2 = np.cos(x)
# plt.plot(x, [y1, y2]) # cannot plot: wrong shape
ys = np.array([y1, y2])
ys = ys.T # transpose of the array ys
plt.plot(x, ys)
plt.title("catch a wave and you're sitting on top of the world!")



```

Random numbers.... (uniform distribution)

```python
random_numbers = np.random.random(10000)

# plt.plot(random_numbers)
plt.hist(random_numbers, rwidth=BAR_CHART_COL_WIDTH)
plt.ylabel("count")
plt.xlabel("value of x")
plt.title("distribution of x values")


```

Random numbers (normal distribution)

```python
n = 10000
z = np.random.normal(size=n)
_ = plt.hist(z, bins=100)
plt.xlabel("number of standard deviations")
plt.ylabel("count")


```

Plotting multiple graphs together

```python
# random numbers from the standard-normal distribution
n = 1000
z = np.random.normal(size=n)
counts, bin_edges, _ = plt.hist(z, bins=100)
plt.xlabel("number of standard deviations")
plt.ylabel("count")
# print("counts = ", counts)
# print("bin_edges = ", bin_edges)

# draw a perfect normal curve
from scipy.stats import norm
x_axis = np.arange(-3, 3, 0.01)
y_axis = norm.pdf(x_axis, 0, 1) # compute the height of the probablity density curve
# correct the scale of pdf values:
bin_width = bin_edges[1] - bin_edges[0] # width of each bin is the same
y_axis *= (n * bin_width)

# plot the histogram and the curve
plt.plot(x_axis, y_axis, color="red")
plt.title("is this normal?")



```

Pie chart

```python
coffee_type = ["coffee black", "iced americano", "hot chocolate with espresso", "mocha"]
counts = [17, 4, 6, 11]

# _ = plt.pie(counts, labels=coffee_type)
# _ = plt.bar(coffee_type, counts)

# create 2 suplots:
fig, ax = plt.subplots(1, 2, figsize=(10,5)) # 1 row, 2 columns, dimensions 10x5
ax[0].pie(counts, labels=coffee_type)
ax[0].set_title("we love pie!")
ax[1].bar(x=coffee_type, height=counts)
ax[1].set_title("what we drink")
plt.xticks(rotation=45, ha="right") # rotate the labels 45 degrees


```
