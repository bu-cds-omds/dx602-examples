Data and variables:
<li>data: values with which we interact in a program
<li>variable: a storage location within the computer's memory that holds some value


```python
# code here
x = 4 # assign the value 4 to a variable called x
y = 13

print('x =', x)
# print(f'x ={x}') # format-string (make substitutions)

text = "hello, world!"
print(text)
y = 42
print('y =', y)

# print('z =', z)
```

Arithmetic Operations

```python
# code here
x = 14
y = 3
# multiplication
m = x * y
print('mult =', m)
# floating point division
d = x / y
print('div =', d)
# integer division: whole number result
id = x // y
print('int div =', id)
# remainder from integer division
r = x % y # modulo operator
print('remainder =', r)

```

Strings and string operations
- concatenation
- indexing
- slicing

```python
# code here
fruit = 'strawberry'
baked_good = 'pie'

print('fruit =', fruit)
print('fruit =', 2 * fruit) # repetition
print(2 * fruit + baked_good) # repetition and concatenation
print(2 * fruit + ' ' + baked_good) # repetition and concatenation
print(2 * (fruit + ' ') + baked_good) # repetition and concatenation

# strings are sequence too... indexing and slicing
print('fruit[0] =', fruit[0]) # extracting the character at position 0
print('fruit[0:5] =', fruit[0:5]) # extracting a substring of 5 character from fruit

```

Lists (a.k.a. "arrays")
<li>collection of data elements (of any type)
<li>contains 0 to many elements; order is maintained
<li>element-access by index (starts at 0)


```python
# declare some lists
numbers = [4,5,6,7,8,9,1,2,3,]
fruits = ['apple', 'banana', 'kiwi', 'strawberry', 'orange', 'grape']

# indexing
# print('numbers[4] =', numbers[4]) # extract the element in position 4
# print('fruits[2] =', fruits[2])
# # negative index numbers:
# print('fruits[-2] =', fruits[-2]) # counting from the end, last elem is -1
# print('fruits[-1] =', fruits[-1]) # counting from the end, last elem is -1

# slicing: obtain a subsequence
print('numbers[0:3] =', numbers[0:3]) # where to start: where to stop before
print('fruits[-3:-1] =', fruits[-3:-1]) # where to start: where to stop before
print('fruits[-1:-4:-1] =', fruits[-1:-4:-1]) # where to start: where to stop before: how to iterate
```

String processing:
<li>Format strings
<li>string slicing by comma
<li>list indexing to extract elements

```python
# format string example:
h = 4
m = 44
s = 31
# # create a string like "04:44:31"
time_string = f'{h:02}:{m:02}:{s:02}' #inside curly braces: any valid expression
print(time_string)

# string parsing example
name_country = 'Brown, Charlie (USA)'
# assumption: country code is 3 characters inside parentheses
# country = name_country[-5:] #
country = name_country[-4:-1] # exclude parentheses
print('country =', country)
name = name_country[:-6] # exclude space and parentheses
print('name =', name)

# string format example:
print(f"name: {name:>20}") # this > format specifier right-aligns
print(f"name: {name:^20}") # this ^ format specifier center-aligns

```
