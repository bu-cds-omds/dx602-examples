Reading a TSV (tab separated values) structured file

```python
# sample TSV file: 
filename = 'data6_a.tsv'
# filename = 'data6_b.tsv'
# filename = 'data6_c.tsv'

with open(filename) as f:

    # strip removes any whitespace
    # split will separate a string into a list of elements
    headers = f.readline().strip().split('\t')
    # print(headers)

    # # the rest of the file is common data records:
    # for line in f:
    #     # print(line.strip().split())
    #     values = line.strip().split()
    #     print(values)

    #     # check for values:
    #     if values:
    #         # build a dictionary of key-value pairs:
    #         d = {}
    #         for i in range(len(headers)):

    #             # check if index i is valid in list values:
    #             if i < len(values):
    #                 # assign the value to this dict with it's key
    #                 d[headers[i]] = int(values[i])

    #         print(d)
    #     else:
    #         print("Encountered empty row.")


    ## using a dictionary comprehension
    for line in f:
        values = line.strip().split('\t')
        d = { headers[i] : values[i] for i in range(len(headers))}
        print(d)

```

Reading a JSON (JavaScript Object Notation) structured file.

```python
# sample json file
filename = 'data6_e.json'

# with open(filename) as f:

#     text = f.read()
#     # print(text)
#     json_data = eval(text) 
#     print(json_data)

# helpful package for reading json format data files
import json
with open(filename) as f:

    json_data = json.load(f) 
    # print(json_data)

    for d in json_data['data']:
        # process one dict at a time
        print(f"unpacking {d}...")

        # go through all k-v pairs in d:
        for k in d:
            print(f"k={k} : v={d[k]}")
```
