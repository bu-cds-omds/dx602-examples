```python
import numpy as np

# create an array of 30 values
x = np.arange(30)
#print('x = ', x)

# look at the shape
#print('x.shape =', x.shape)

x = x.reshape((5,6)) # 5x6 is 30
print('x = ')
print(x) # new shape, same elements

# slice rows:
print("x[2:4] =")
print(x[2:4])
# slice columns
print('x[:,3:5] = ')
print(x[:,3:5]) # slice all rows, columns 3 stop before 5

# slice rows and columns together:
print('x[2:4, 3:5] = ')
print(x[2:4, 3:5])
```

Iterating through columns

```python
#a = np.arange(1, 21).reshape((5,7)) # cannot reshape array of size 20 into shape (5,7)
a = np.arange(1, 21).reshape((4,5))
print(a)
# if we attempt to slice with indexes out of range
# we will get the largest possible slice, no error
#print(a[2:12, 3:8])

# loop through rows:
# for i in range(len(a)):
for i in range(a.shape[0]): # a.shape[0] is the number of rows
    # print the row at index i
    print(f'row {i} = {a[i]}')

# loop through the columns
for j in range(a.shape[1]): # a.shape[1] is the number of cols
    # print each column:
    print(f'col {j} = {a[:,j]}') # slice all rows, col j

# show an example of printing a sliced column top-down
c = a[:,1] # all rows, column 1
# print('c = \n', c.reshape((4,1)))
# print('c = \n', c.reshape((len(c),1)))
print('c = \n', c.reshape((-1,1)))

```

Linear algebra: example weighted average

```python
# consider a set of 3 loans, e.g. car loan; mortgage; credit card
loans = np.array([34000, 327000, 23456]) 
rates = np.array([0.075, 0.052, 0.185])

print(f'loans = {loans}')
print(f'rates = {rates}')

# how much interest on each loan (per year, in dollars)?
interest = loans * rates
print(f'interest = {interest}')
print(f'total interest: {interest.sum()}')

# what is the percentage of total debt in each loan?
weights = loans / loans.sum()
print(f'weights = {weights}')

# WACC: weighted-average cost of capital
wacc = weights @ rates # dot-product multiplication
print(f'wacc = {wacc:.03}')
```

Monte-carlo simulation: estimate the value of pi

```python
import random # plain python random library
import time

nthrows = 10000000
hits = 0 # count how many times land inside the circle
# # (1) the plain python way:
print("the plain old python way:")

bt = time.time() # begin time

for i in range(1, nthrows+1):
    # pick some random coordinates for (x,y)
    x = random.random() # uniformly distributed btw (0,1)
    y = random.random() # uniformly distributed btw (0,1)

    # print(f'throw {i}: ({x:0.2f}, {y:0.2f})')
    if (x ** 2 + y ** 2) <= 1:
        # print("It's a hit! (inside the circle)")
        hits += 1 

    estpi = 4 * hits / i # number thrown

et = time.time() # end time
print(f'{hits} out of {i} throws, estpi = {estpi:.6f}, time={et-bt:.4f} seconds')

# (2) the numpy way
print("the numpy way")
bt = time.time() # begin time
x = np.random.random(nthrows)
y = np.random.random(nthrows)

# count the hits:
hits = ((x ** 2) + (y ** 2) <= 1).sum()
estpi = 4 * hits / nthrows
et = time.time() # end time
print(f'hits = {hits} out of {nthrows}, estpi = {estpi:.6f}, time={et-bt:.4f} seconds')
```
